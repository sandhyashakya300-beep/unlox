```python
'''Q17. Swiggy Order Calculator. Take item price and quantity as input. Compute total = price × 
quantity. Add 18% GST. Add ₹40 delivery charge if total (before delivery) is below ₹500, else delivery 
is FREE. Print all components separately and final amount in this format:
Subtotal: ₹450.00
GST (18%): ₹81.00
Delivery: ₹40.00
------------------------
Total: ₹571.00'''

# Take inputs from the user
price = float(input("Enter the item price: "))
quantity = int(input("Enter the quantity: "))

# Calculate the basic cost 
subtotal = price * quantity

# Calculate 18% GST on the subtotal
gst = subtotal * 0.18

# Delivery is ₹40 if the subtotal is under ₹500, otherwise it's free (₹0)
if subtotal < 500:
    delivery = 40.0
else:
    delivery = 0.0

# Calculate the final total sum
final_total = subtotal + gst + delivery

# Print the receipt 
print(f"Subtotal: ₹{subtotal:.2f}")
print(f"GST (18%): ₹{gst:.2f}")
print(f"Delivery: ₹{delivery:.2f}")
print(f"Total: ₹{final_total:.2f}")
```

```python
'''Q21. Indian Income Tax Calculator (New Regime, FY 2024-25). Take annual income as input. 
Compute tax using these slabs:
Up to ₹3,00,000 : Nil
₹3,00,001 - ₹7,00,000 : 5%
₹7,00,001 -₹10,00,000 : 10%
₹10,00,001 - ₹12,00,000 : 15%
₹12,00,001 -₹15,00,000 : 20%
Above ₹15,00,000 : 30%
Tax must be computed slab-wise, NOT flat. Example: income ₹8,00,000 → 0 on first 3L + 5% on next 4L + 10% on 
last 1L = ₹20,000 + ₹10,000 = ₹30,000. Print breakdown clearly.'''
income = float(input("Enter your annual income: ₹"))
total_tax = 0

print("\n--- Tax Breakdown ---")

# Slab 1: Up to ₹3,00,000 (Nil)
if income > 0:
    amount = min(income, 300000) 
    print(f"0% tax on first ₹{amount}: ₹0")

# Slab 2: ₹3,00,001 to ₹7,00,000 (5%)
if income > 300000:
    amount = min(income - 300000, 400000)
    tax = amount * 0.05
    total_tax += tax
    print(f"5% tax on next ₹{amount}: ₹{tax}")

# Slab 3: ₹7,00,001 to ₹10,00,000 (10%)
if income > 700000:
    amount = min(income - 700000, 300000)
    tax = amount * 0.10
    total_tax += tax
    print(f"10% tax on next ₹{amount}: ₹{tax}")

# Slab 4: ₹10,00,001 to ₹12,00,000 (15%)
if income > 1000000:
    amount = min(income - 1000000, 200000)
    tax = amount * 0.15
    total_tax += tax
    print(f"15% tax on next ₹{amount}: ₹{tax}")

# Slab 5: ₹12,00,001 to ₹15,00,000 (20%)
if income > 1200000:
    amount = min(income - 1200000, 300000)
    tax = amount * 0.20
    total_tax += tax
    print(f"20% tax on next ₹{amount}: ₹{tax}")

# Slab 6: Above ₹15,00,000 (30%)
if income > 1500000:
    amount = income - 1500000
    tax = amount * 0.30
    total_tax += tax
    print(f"30% tax on remaining ₹{amount}: ₹{tax}")
print(f"Total Income Tax to pay: ₹{total_tax}")
```

```python
'''Q24. Cricket Strike Rate Classifier. Take runs scored and balls faced. Compute strike rate = 
(runs/balls) x 100. Classify:
SR > 150 → Explosive (T20 finisher)
100 < SR ≤ 150 → Aggressive (Top order T20)
75 ≤ SR ≤ 100 → Steady (ODI anchor)
SR < 75 → Defensive (Test match)
Edge case: handle balls = 0 (do not crash, print message instead).'''
runs = int(input("Enter runs scored: "))
balls = int(input("Enter balls faced: "))
if balls == 0:
    print("Cannot calculate strike rate: batsman hasn't faced any balls yet!")
else:
    # Calculate the strike rate
    sr = (runs / balls) * 100
    print(f"Strike Rate: {sr:.2f}")
    
    # Classify based on the SR
    if sr > 150:
        print("Classification: Explosive (T20 finisher)")
    elif sr > 100:  
        print("Classification: Aggressive (Top order T20)")
    elif sr >= 75:
        print("Classification: Steady (ODI anchor)")
    else:
        print("Classification: Defensive (Test match)")
```

```python
'''Q29. Print the first 5 rows of Pascal's Triangle:
 1
 1 1
 1 2 1
 1 3 3 1
 1 4 6 4 1
You may use any approach (formula-based or direct calculation), but no math.comb() or any imported function.'''
n = 5
current_row = [1]

for i in range(n):
    for num in current_row:
        print(num, end=" ")
    print() 
    
    # Step 2: Build the next row to use for the next loop
    next_row = [1] 
    
    # Add the pairs of numbers from the current row together
    for j in range(len(current_row) - 1):
        added_value = current_row[j] + current_row[j+1]
        next_row.append(added_value)
        
    next_row.append(1) # Every row always ends with a 1
    
    #  Replace the old row with our newly built row
    current_row = next_row
```

```python
'''Q31. Find all Armstrong numbers between 1 and 9999. (An n-digit number is Armstrong if the sum of 
its digits each raised to the power n equals the number itself. Example: 153 = 1³ + 5³ + 3³.) Print them 
in a clean comma-separated list with a count at the end.'''
armstrong_list = []

# Loop through all the numbers from 1 to 9999
for num in range(1, 10000):
    
    # Convert to string to easily find the length (n) and grab digits
    num_str = str(num)
    n = len(num_str)
    
    total_sum = 0
    for digit in num_str:
        total_sum += int(digit) ** n
        
    # If the math adds up to the original number, it's an Armstrong!
    if total_sum == num:
        armstrong_list.append(str(num))
print(", ".join(armstrong_list))
print(f"Total count: {len(armstrong_list)}")
```

```python
'''Q34. Take a number as input. Reverse it WITHOUT converting to string. Then check if the original and 
reversed are equal (palindrome). Handle negative numbers, they should be considered NOT 
palindrome (because of the minus sign). Examples: 121 → palindrome, 123 → not palindrome, -121 
→ not palindrome.'''
num = int(input("Enter a number: "))

if num < 0:
    print(f"{num} -> not palindrome")
else:
    
    original_num = num
    reversed_num = 0
    
    
    while num > 0:
    
        last_digit = num % 10
        reversed_num = (reversed_num * 10) + last_digit
        num = num // 10
    if original_num == reversed_num:
        print(f"{original_num} -> palindrome")
    else:
        print(f"{original_num} -> not palindrome")
```

```python
'''Q35. Take two positive integers. Compute their HCF (Highest Common Factor) using Euclidean 
algorithm with a while loop. Then compute LCM using the relation: HCF x LCM = a x b. Do not use 
math.gcd.'''
# Take the two numbers from the user
num1 = int(input("Enter the first positive number: "))
num2 = int(input("Enter the second positive number: "))
a = num1
b = num2

while b > 0:
    remainder = a % b
    a = b
    b = remainder
hcf = a
lcm = (num1 * num2) // hcf

print(f"\nHCF of {num1} and {num2} is: {hcf}")
print(f"LCM of {num1} and {num2} is: {lcm}")
```

```python
'''Q36. String scanner, single pass. Take a sentence (e.g., a Swiggy review like "Order came on time, 
food was hot, packaging 10/10!"). In ONE pass through the string, count: total characters, total 
words, vowels (a, e, i, o, u - case-insensitive), consonants, digits, spaces, and special characters. Print 
all counts.'''
text = input("Enter a sentence: ")

# Set all our counters to zero at the start
total_chars = 0
vowels = 0
consonants = 0
digits = 0
spaces = 0
special_chars = 0
words = 0

# A flag to help us count words without using .split()
in_a_word = False 

# Loop through the string exactly one time character by character
for char in text:
    total_chars += 1
    
    if char != ' ' and not in_a_word:
        words += 1
        in_a_word = True
    elif char == ' ':
        in_a_word = False
        
    if char == ' ':
        spaces += 1
    elif char.isdigit():
        digits += 1
    elif char.isalpha():
        # .lower() makes it case-insensitive so 'A' and 'a' both count
        if char.lower() in 'aeiou':
            vowels += 1
        else:
            consonants += 1
    else:
        
        special_chars += 1

# Print the final report
print("\n--- String Scanner Results ---")
print(f"Total Characters: {total_chars}")
print(f"Total Words: {words}")
print(f"Vowels: {vowels}")
print(f"Consonants: {consonants}")
print(f"Digits: {digits}")
print(f"Spaces: {spaces}")
print(f"Special Characters: {special_chars}")
```

```python
'''Q37. Take 10 student marks (out of 100) as input, space-separated on one line, parse with split() and 
map(). Without using sort() or sorted(), find: (a) maximum, (b) minimum, (c) average, (d) count of 
students above average, (e) the SECOND highest mark.'''
# Take input and convert it to a list of integers using split() and map()
marks_input = input("Enter 10 student marks separated by spaces: ")
marks = list(map(int, marks_input.split()))

max_mark = -1
second_max = -1
min_mark = 101  
total_sum = 0

# Loop through the marks to find the max, min, sum, and second max
for mark in marks:
    total_sum += mark
    
    # Check for the minimum mark
    if mark < min_mark:
        min_mark = mark
        
    # Check for the maximum and second maximum
    if mark > max_mark:
        second_max = max_mark
        max_mark = mark
        
    elif mark > second_max and mark != max_mark:
        second_max = mark

#  Calculate the average
average = total_sum / len(marks)

#  Loop one more time just to count who scored strictly above average
above_average_count = 0
for mark in marks:
    if mark > average:
        above_average_count += 1
print(f"(a) Maximum mark: {max_mark}")
print(f"(b) Minimum mark: {min_mark}")
print(f"(c) Average mark: {average}")
print(f"(d) Students above average: {above_average_count}")
print(f"(e) Second highest mark: {second_max}")
```

```python
'''Q38. Word Frequency Analyzer for Swiggy reviews. Take a review sentence. Find the frequency of 
each word (case-insensitive, "Food" and "food" should be counted together). Print words sorted by 
frequency in descending order. Also print the MOST frequent word and its count. Use a dict.'''

review = input("Enter your Swiggy review: ")

# Make everything lowercase so "Food" and "food" are treated the same
review = review.lower()

# Split the sentence into individual words
words = review.split()

# Build the dictionary to count the frequencies
word_freq = {}

for word in words:
    if word in word_freq:
        word_freq[word] += 1  
    else:
        word_freq[word] = 1 

#  Sort the dictionary. 

sortable_list = []
for word, count in word_freq.items():
    sortable_list.append((count, word))

# Sort in descending order (highest numbers at the top)
sortable_list.sort(reverse=True)

for count, word in sortable_list:
    print(f"{word}: {count}")

if len(sortable_list) > 0:
    top_count = sortable_list[0][0]
    top_word = sortable_list[0][1]
    print(f"\nMOST frequent word: '{top_word}' (used {top_count} times)")
```

```python
'''Q39. Indian States CRUD. Pre-populate a dict with 5 Indian states and their capitals (e.g., Karnataka 
→ Bengaluru, Maharashtra → Mumbai). Then build a menu-driven program:
1. Add a new state-capital pair
2. Update an existing capital
3. Delete a state
4. Search a state (return capital, or 'Not found')
5. Display all states sorted alphabetically
6. Exit
Use a while loop. Loop until user chooses option 6.'''
# Starting out with our 5 pre-populated states
india_states = {
    "Maharashtra": "Mumbai",
    "Karnataka": "Bengaluru",
    "Tamil Nadu": "Chennai",
    "Uttar Pradesh": "Lucknow",
    "Gujarat": "Gandhinagar"
}

while True:
    print("1. Add a new state-capital pair")
    print("2. Update an existing capital")
    print("3. Delete a state")
    print("4. Search a state")
    print("5. Display all states sorted alphabetically")
    print("6. Exit")
    
    choice = input("Enter your choice (1-6): ")
    
    if choice == '1':
        state = input("Enter state name: ").title() 
        capital = input("Enter capital name: ").title()
        india_states[state] = capital
        print(f"Added {state} -> {capital} successfully.")
        
    elif choice == '2':
        state = input("Enter the state you want to update: ").title()
        if state in india_states:
            new_capital = input(f"Enter the new capital for {state}: ").title()
            india_states[state] = new_capital
            print("Capital updated successfully!")
        else:
            print("Error: State not found.")
            
    elif choice == '3':
        state = input("Enter the state to delete: ").title()
        if state in india_states:
            del india_states[state]
            print(f"{state} has been deleted.")
        else:
            print("Error: State not found.")
            
    elif choice == '4':
        state = input("Enter state to search: ").title()
        result = india_states.get(state, "Not found")
        print(f"Search Result: {result}")
        
    elif choice == '5':
        for state in sorted(india_states.keys()):
            print(f"{state} -> {india_states[state]}")
            
    elif choice == '6':
        print("Exiting program. Goodbye!")
        break  
        
    else:
        print("Invalid choice! Please pick a number from 1 to 6.")
```

```python
'''Q40. Set Operations. Students in Section A know languages = {Python, Java, C++, SQL}. Students in 
Section B know = {Java, JavaScript, Python, Go, SQL}. Using ONLY set operations, find:
(a) Languages known by BOTH sections
(b) Languages known ONLY by Section A
(c) Languages known ONLY by Section B
(d) Total unique languages across both sections
(e) Languages exclusive to one section but not both (symmetric difference)'''

section_a = {"Python", "Java", "C++", "SQL"}
section_b = {"Java", "JavaScript", "Python", "Go", "SQL"}

both = section_a & section_b
print(f"(a) Known by BOTH: {both}")
only_a = section_a - section_b
print(f"(b) ONLY in Section A: {only_a}")


only_b = section_b - section_a
print(f"(c) ONLY in Section B: {only_b}")

total_unique = section_a | section_b
print(f"(d) Total unique across both: {total_unique}")
exclusive = section_a ^ section_b
print(f"(e) Exclusive to one section but not both: {exclusive}")
```

```python
'''Q41. TCS stock prices for a week stored as a tuple: `prices = (3845, 3902, 3878, 3955, 4012, 3998, 
4055)` (Mon to Sun in ₹). Find: (a) maximum price and which day, (b) minimum price and which day, 
(c) average, (d) biggest single-day gain (and between which days), (e) total volatility = sum of 
absolute day-to-day price changes.'''
# The given tuple of prices
prices = (3845, 3902, 3878, 3955, 4012, 3998, 4055)

# A list of days to match up with the prices
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
max_price = max(prices)
max_day_index = prices.index(max_price)
max_day = days[max_day_index]

# (b) Minimum price and day
min_price = min(prices)
min_day_index = prices.index(min_price)
min_day = days[min_day_index]

average_price = sum(prices) / len(prices)

max_gain = 0
best_days = ""
total_volatility = 0

for i in range(len(prices) - 1):
    price_change = prices[i+1] - prices[i]
    total_volatility += abs(price_change)
    
    if price_change > max_gain:
        max_gain = price_change
        best_days = f"{days[i]} and {days[i+1]}"

# Print all the final answers 
print("--- TCS Stock Analysis ---")
print(f"(a) Maximum price: ₹{max_price} on {max_day}")
print(f"(b) Minimum price: ₹{min_price} on {min_day}")
print(f"(c) Average price: ₹{average_price:.2f}")
print(f"(d) Biggest single-day gain: ₹{max_gain} (between {best_days})")
print(f"(e) Total volatility: ₹{total_volatility}")
```

```python
'''Q42. Given a list of 20 integers (you generate or hardcode them, some duplicates), produce: (a) the 
set of unique numbers, (b) the list of numbers appearing more than once, (c) the list of numbers 
appearing EXACTLY once. Do this WITHOUT importing Counter or any other library.'''

numbers = [5, 2, 8, 5, 9, 2, 1, 4, 8, 8, 3, 7, 6, 9, 10, 15, 12, 1, 5, 20]

unique_numbers = set(numbers)

#  Count how many times each number appears using a dictionary
number_counts = {}

for num in numbers:
    if num in number_counts:
        number_counts[num] += 1  
    else:
        number_counts[num] = 1   

#  Separate them into two different lists based on their counts
more_than_once = []
exactly_once = []

for num, count in number_counts.items():
    if count > 1:
        more_than_once.append(num)
    elif count == 1:
        exactly_once.append(num)

# Print all the final results
print(f"Original list (20 items): {numbers}\n")

print(f"(a) Set of unique numbers: {unique_numbers}")
print(f"(b) Numbers appearing MORE than once: {more_than_once}")
print(f"(c) Numbers appearing EXACTLY once: {exactly_once}")
```

```python
'''Q43. Smart Palindrome Checker. Take a sentence as input. Check if it is a palindrome IGNORING: 
case, spaces, and all special characters (punctuation). Examples that should return TRUE: "A man, a 
plan, a canal: Panama", "Race car", "No 'x' in Nixon". Hint: build a cleaned version of the string first.'''
sentence = input("Enter a sentence to check: ")

cleaned_string = ""

for char in sentence:
    # .isalnum() checks if the character is a letter or a number.
    # It automatically skips over any spaces, commas, or punctuation!
    if char.isalnum():
        cleaned_string += char.lower()

#  Check if the clean string is exactly the same forwards and backwards
# [::-1] is the standard Python trick to quickly reverse a string
if cleaned_string == cleaned_string[::-1]:
    print("TRUE")
else:
    print("FALSE")
```

```python
'''Q44. Manual Title Case. Take a sentence (e.g., "hello world from bengaluru"). Capitalize the first 
letter of EACH word, but do NOT use .title() or .capitalize(). Use only .upper() / .lower() and string 
indexing. Output: "Hello World From Bengaluru".'''
sentence = input("Enter a sentence: ")

# Split the sentence into a list 
words = sentence.split()
title_cased_words = []

for word in words:
    #  make it uppercase
    first_letter = word[0].upper()
    rest_of_word = word[1:].lower()
    

    new_word = first_letter + rest_of_word
    title_cased_words.append(new_word)

final_result = " ".join(title_cased_words)

print(f'Output: "{final_result}"')
```

```python
'''Q45. Caesar Cipher Encryptor. Take a message and a shift value N as input. Encrypt by shifting each 
letter forward by N positions in the alphabet (wrapping around, Z + 1 = A). Preserve case. Keep digits, 
spaces, and special characters UNCHANGED. Example: "Hello, India!" with shift 3 → "Khoor, Lqgld!"'''
message = input("Enter the message: ")
shift = int(input("Enter shift value (N): "))

encrypted_message = ""

for char in message:
    # Handle UPPERCASE letters
    if char.isupper():
        shifted_value = (ord(char) - 65 + shift) % 26
        new_char = chr(shifted_value + 65)
        encrypted_message += new_char
        
    # Handle lowercase letters
    elif char.islower():
        shifted_value = (ord(char) - 97 + shift) % 26
        new_char = chr(shifted_value + 97)
        encrypted_message += new_char
        
    else:
        encrypted_message += char

print(f'Output: "{encrypted_message}"')
```

```python
'''Q46. Write a function calculate_gst(amount, gst_rate=18, gst_inclusive=False):
- If gst_inclusive=False: amount is base, add GST on top
- If gst_inclusive=True: amount already includes GST, extract it
- Return a tuple: (base_amount, gst_amount, total_amount)
- All values rounded to 2 decimals
Test cases:
 calculate_gst(1000) → (1000.00, 180.00, 1180.00)
 calculate_gst(1180, gst_inclusive=True) → (1000.00, 180.00, 1180.00)
 calculate_gst(500, gst_rate=5) → (500.00, 25.00, 525.00)'''
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    
    #  GST is NOT included yet (we need to add it)
    if gst_inclusive == False:
        base_amount = amount
        gst_amount = amount * (gst_rate / 100)
        total_amount = base_amount + gst_amount
        
    # GST is ALREADY included (we need to extract it)
    else:
        total_amount = amount
        base_amount = amount / (1 + (gst_rate / 100))
        gst_amount = total_amount - base_amount
        
    
    return (round(base_amount, 2), round(gst_amount, 2), round(total_amount, 2))

# --- Testing the exact cases from the question ---
print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
```
