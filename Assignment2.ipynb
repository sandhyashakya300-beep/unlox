Q1 : Flow of execution : explain the flow of execution in a python program that contains functions , loops and conditional statements.
     
     ANSWERS : The flow of execution is the specific order in which a computer evaluates and runs lines of code. By default, Python reads code sequentially, from top to bottom. However, structural elements completely alter this linear path:
      
      ​Functions: When Python encounters a function definition (def), it simply memorizes the function's name and code block but does not execute it. The flow skips over the body of the function entirely until it sees a function call. When called, the main execution pauses, jumps inside the function, runs those lines, and then jumps right back to where it left off.
​       
      Conditional Statements (if/elif/else):  Python evaluates a boolean condition. If it is True, the execution enters that specific indented block and completely bypasses the alternative elif or else branches.
      
      ​Loops (for/while):  Instead of moving down to the next line of code, the execution reaches the end of the loop block and jumps back up to the loop's header and move to the next item in a sequence.

Q4 : Internal mechanics of loops: What happens internally when a for loop uses the range()function ? explain with an example.
       
       ANSWERS :When you write a loop like for i in range(3):, Python does not actually create a static list containing [0, 1, 2] in memory. Doing so would be incredibly inefficient if you ran range(1000000).

               Instead, range() generates a lazy, immutable iterable sequence object. It only knows three things: the start value, the stop value, and the step size.

               Python evaluates range(3) and creates the range object.
                 The for loop implicitly calls Python's internal iter() function on this object, which creates an iterator. This iterator keeps track of a pointer (current state).
                 At the start of every iteration, the loop calls an internal function named _next_() on that iterator.
                 The iterator calculates the next number on the fly, updates its internal pointer, and hands that number to the loop variable i.
                 When the range is exhausted, the iterator raises a special, hidden exception called StopIteration. Python intercepts this exception and cleanly terminates the loop without crashing.

Q6 : Loop control : what is the importance of loop control conditions ? what issues may arise if loop conditions are not defined properly?

        ANSWER : Loop control conditions define the exact boundaries of a loop's lifetime. Their core importance lies in ensuring           predictability: defining precisely when a task is "done." Without properly defined exit boundaries, loops become erratic and dangerous.
         
         Issues from Improper Design
         
          Infinite Loops: If the control condition never evaluates to False, the loop runs forever. For example, in a while count < 5: loop, if you forget to increment count inside the block, the condition stays True indefinitely. This spikes CPU utilization to 100%, freezes user interfaces, and eventually crashes the application or system.
          
          Off-by-One Errors: This occurs when a loop runs either one time too many or one time too few. It usually stems from confusing < with <=. In an array of size 5, trying to access index 5  will trigger an IndexError because indices only go from 0 to 4.
          
          Premature Termination: If internal loop logic updates a condition too early, or triggers a break statement mistakenly, the loop exits before processing all necessary data, leading to incomplete computations.

Q7 : Numpy vs lists : explain how NumPy arrays differ from python lists in terms of memory usage and performance.

         ANSWER : While both store collections of elements, NumPy arrays (ndarrays) and Python lists are engineered completely differently .
             
             PYTHON LISTS : --> array of pointers scattered across memory .
                           ---> heterogeneous data .
                           ---> slow processing speed.

            NUMPY ARRAY : ---> single uninterested block of raw memory .
                          ---> strictly homogeneous data .
                          ---> fast processing speed .

          Memory and Performance Mechanics
​               A Python list doesn't actually hold values; it holds references (pointers) to objects scattered elsewhere in memory. This           introduces heavy overhead because each element requires its own object tracking wrapper, and accessing elements causes          cache misses (the CPU has to constantly hunt around memory locations).
                ​NumPy arrays store raw, unboxed data back-to-back in a single physical block of memory. Because the data type is uniform, NumPy bypasses Python's dynamic type-checking entirely. It performs calculations via vectorization—applying an operation to the entire array simultaneously at the CPU hardware level using SIMD (Single Instruction, Multiple Data) architecture, making operations orders of magnitude faster than a standard Python list loop

```python
#9.Sum of digits : Write a program that takes a numbers from the users , uses a loop to calculate the sum of its digits and display the results.
#take input from the user
num = int(input("Enter the number:"))
digit_sum = 0
#use while loop for looping 
while num > 0:
    last_digit = num % 10
    digit_sum = digit_sum + last_digit
    
    num = num //10
#display the results 
print(f"the sum of digits is {digit_sum}")
```

```python
#10.Even filter function: Write a function that accepts a list of integers and return a new list containing only even numbers.

def filter_even_num(num):
    #initialized an empty list
    even_list = []

    for number in num:
        if number % 2 == 0:
            even_list.append(number)
    return even_list

#list 
sample_num = [11,22,13,67,87,23,34,43,54,56,99,76]
result = filter_even_num(sample_num)
print(f"original list is{sample_num}")
print(f"filtered even number list is {result}")
```

```python
#12.List reversal : write a program that takes a list of numbers from the users, uses slicing to reverse the list  , and 
#print both the original and reversed list.

#take user input 
user_input = input("enter the number separated by space :")
original_list = user_input.split()

for i in range(len(original_list)):
    original_list[i] = int(original_list[i])

#using slicing to reverse the list 
reversed_list = original_list[::-1]

#display both the lists
print(f"the original list is {original_list}")
print(f"the reversed list is {reversed_list}")
```

```python
#13.Numpy slicing : create a Numpy array with atleast 10 element and write a program to:
#extract the first half of the array slicing 
#calculate the sum of the extracted elments.

import numpy as np
#create an array
my_array = np.array([10,22,12,33,21,34,56,54,57,98,87])
print(f"the original numpy array is {my_array}")

midpoint = len(my_array) // 2

#extract the first half of the array slicing
first_half = my_array[0:midpoint]
print(f"first half of the array is {first_half}")

#calculate the sum of the extracted elements 
array_sum = np.sum(first_half)
print(f"sum of the extracted elements is {array_sum}")
```

```python
#14.Arithmetic palindrome: write a program to check whether a given number is a palindrome using arithmetic operations.

#take a inpit from the user 
og_num = int(input("Enter a number to check for palindrome:"))
temp = og_num
reversed_num = 0
#looping until all digits have been extracted
while temp > 0:
    last_digit = temp % 10
    reversed_num = (reversed_num*10) + last_digit
    temp = temp // 10

if og_num == reversed_num:
    print(f"original number is {og_num}  a palindrome")
else:
    print(f"original number {og_num} is not a palindrome")
```
