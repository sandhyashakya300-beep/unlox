```python
#handling missing value
#adjust these columns based on csv file
df.dropna(inplace=True)
```

```python
#data operations 
if 'Order_date' in df.columns :
    df['Order_date'] = pd.to_datetime(df['Order_date'])

print("\nData preprocessing complete :")
print(df.head(4))
```

```python
total_revenue = df['Revenue'].sum()
total_profit = df['Profit'].sum()
total_quantity = df['Quantity'].sum()
avg_profit_margin = (total_profit / total_revenue)*100
print(f"total unit sold:{total_quantity:,} ")
print(f"total revenue: ${total_revenue:,.2f}")
print(f"total net profit:${total_profit}")
print(f"average profit margin : {avg_profit_margin}%")
```

3. EXPLORATORY DATA ANALYSIS (EDA)

```python
regional_summary = df.groupby('Region')[['Revenue','Profit']].sum().reset_index()
print(regional_summary)
category_summary = df.groupby('Category')[['Revenue','Profit']].sum().sort_values(by='Revenue',ascending=False)
print(category_summary)
top_product = df.groupby('Product_Name')[['Quantity','Revenue','Profit']].sum().sort_values(by='Profit',ascending=False).head(11).reset_index()
print(top_product)
```

```python
#time-series trend analysis
print("\n--monthly revenue trends---")
df['Year_Month'] = df['Order_Date']
monthly_trend = df.groupby('Year_Month')[['Revenue','Profit']].sum().reset_index()
monthly_trend['Year_Month'] = monthly_trend['Year_Month'].astype(str)
print(monthly_trend)
```

```python
# 1. Define the variables correctly
date_col = 'Order_Date'
expense_col = 'Revenue'
category_col = 'Category'

# 2. Convert to datetime immediately in the same cell
df[date_col] = pd.to_datetime(df[date_col])
daily_expenses = df.groupby(df[date_col].dt.date)[expense_col].sum()
```

VISUALIZATION 

```python
#Line plot for expense trends

plt.figure(figsize=(11,4))
# Group by the date to track the daily accumulation of expenses
daily_expenses = df.groupby(df[date_col].dt.date)[expense_col].sum()

# Plotting the trend line
plt.plot(daily_expenses.index, daily_expenses.values, color='blue', linewidth=2, markersize=4)

plt.title('Daily Expense Trends Over Time', fontsize=13, pad=12, fontweight='bold')
plt.xlabel('Date', fontsize=11)
plt.ylabel('Total Expenses ($)', fontsize=11)

plt.tight_layout()
plt.show()
```

```python
#Bar chart for category - wise spending

plt.figure(figsize=(10,7))
category_spending = df.groupby(category_col)[expense_col].sum().sort_values(ascending=False)
sns.barplot(x=category_spending.values, y=category_spending.index, palette='pink_r',hue=category_spending.index,legend=False)
plt.title("total spending distribution across categories",fontsize=12,fontweight='bold')
plt.xlabel("total amount spend")
plt.ylabel("expense category")
plt.tight_layout()
plt.show()
```

```python
#Box plot expense variability
plt.figure(figsize=(8, 5))

# Box plot helps identify the median distribution and individual outliers
sns.boxplot(x=df[expense_col], color='green', width=0.4)

plt.title('Expense Spreadsheet Variability & Outlier Check', fontsize=13, pad=12, fontweight='bold')
plt.xlabel('Transaction Amounts ($)', fontsize=11)

plt.tight_layout()
plt.show()
```
