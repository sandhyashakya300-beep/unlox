```python
df['BHK'] = df['size'].str.split(' ').str[0]
```

```python
def sqft_to_num(x):
    tokens = x.split('-')
    if len(tokens) == 2:
        return (float(tokens[0]) + float(tokens[1])) / 2
    try:
        return float(x)
    except:
        return None
    
df['total_sqft'] = df['total_sqft'].apply(sqft_to_num)
```

```python
df.dtypes
```

```python
df['bath'] = df['bath'].fillna(df['bath'].median())
```

```python
df['balcony'] = df['balcony'].fillna(df['balcony'].median())
```

```python
df['BHK'] = pd.to_numeric(df['BHK'])
```

```python
df['BHK'] = df ['BHK'].fillna(df['BHK'].median())
```

```python
df = df.dropna()
```

```python
df['area_type'].unique()    
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()
df['encoded_area'] = encoder.fit_transform(df['area_type'])
df.head(9)
```

```python
encode_df = pd.get_dummies(df['area_type'], prefix='area')
df = pd.concat([df, encode_df], axis=1)
```
