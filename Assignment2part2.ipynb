1. Primary keys: Why is it essential for every table to have a Primary Kry? List three characteristic that a columns must have to 
    qualify as Primary key.

      ANSWER : A Primary Key is essential because it uniquely identifies each record in a table,
                ensuring no duplicate rows exist and    making data retrieval fast and accurate. Without it, finding or updating a specific row becomes unreliable. To qualify as a Primary Key, a column must have these three characteristics:
               
               Uniqueness: Every value in that column must be completely unique.
               Not Null: It cannot contain empty or missing (NULL) values.
               Immutability: The values should rarely, if ever, change over time

2. The Purpose of SQL:MYSQL use SQL to interact with data.Differentiate between Data Definition Language(DDL) and Data Manipulation 
     Language(DML) with one example of a command for each.

       ANSWER : Data Definition Language (DDL) is used to define, alter, or destroy the structure of database objects like tables 
               and schemas. 
               An example of a DDL command is CREATE TABLE.
               Data Manipulation Language (DML) is used to manage, modify, and retrieve the actual data stored within those structures. An example of a DML command is INSERT INTO.

3.Database Normalization : What is the main goal of normalizing a database? Briefly explain what "Data Redundancy " is and why it can be
     problematic for a system.

       ANSWER : The main goal of normalization is to eliminate data redundancy and ensure data dependencies make logical sense.
                 Data Redundancy refers to storing the same piece of information in multiple places across a database. This is highly problematic because it wastes storage space and leads to "data anomalies," where updating a piece of info in one place but forgetting to update it elsewhere causes conflicting, inaccurate data.

4. Foreign key and Relationships: Describe the role of a Foreign Key .How does it help in maintaining "Referential Integrity" between two 
     tables?

        ANSWER :The role of a Foreign Key is to establish and enforce a link between the data in two tables,
                acting as a bridge from one table to another. It maintains Referential Integrity by ensuring that the value in the Foreign Key column must always match an existing Primary Key value in the parent table. This prevents accidental data corruption, like deleting a record that other tables are still actively relying on

5. Understanding One-to-Many Relationships : Provide a real-world example of a One-to-Many relationship(e.g.,Customer and Order ). 
     Which tables should hold the Foreign Key in this scenario , and why?

       ANSWER : A real-world example is an Author and Books relationship, where one author can write many books, 
                but each book has only one author. In this scenario, the Books table (the "Many" side) should hold the Foreign Key pointing to the Author's ID. This is because a single book only needs one slot to store its author's ID, whereas putting it in the Author table would require a messy, infinite list of book IDs in a single cell.

6. The Concepts of Relational Databases: What makes a database "relational"? Explain how data is organized in this model compared to a  
      simple spreadsheet.

         ANSWER : A database is "relational" because it organizes data into distinct, normalized tables 
                  that connect to one another through common fields (keys). A simple spreadsheet usually stores everything in one giant, flat sheet, leading to massive repetition and messy data. Relational databases solve this by splitting data logically into separate tables (like Customers, Products, and Orders), which cuts down on clutter and keeps everything synchronize

7. Data Types : Why is it important to choose the correct data types (e.g.,INT VS VARCHAR VS DECIMAL ) when designing a table ? What could 
        happen if you store a phone number as an INT?

          ANSWER : Choosing the correct data type ensures data accuracy, optimizes storage space,
                  and allows the system to perform proper operations on the data. If you store a phone number as an INT (Integer), several things will break: leading zeros (like the '0' in a mobile prefix or zip code) will be dropped completely, the number might exceed the maximum allowed size for a standard integer, and you won't be able to format it cleanly with hyphens or country codes.

8. NULL Values : In MYSQL , what does a NULL values represent? Is it same as a zero (0) or an empty string (")?. explain your answer.

      ANSWER : In MySQL, a NULL value represents missing, unknown, or inapplicable data.  
               It is fundamentally different from a zero (0) or an empty string (""). A zero is a specific numerical value, and an empty string is a text field with a length of zero characters; both represent known data. NULL, on the other hand, means no data has been entered at all—it is simply a placeholder for the absence of a value.


9. The Joys of Joining : Why do we split data into multiple tables only to "Join" them later?. Explain the basic difference between an 
      INNER JOIN and a LEFT JOIN .

         ANSWER : We split data into multiple tables to keep it clean, normalized, and free of redundant clutter,  
                  using "Joins" only when we need to reconstruct the full picture for a query. An INNER JOIN only returns rows where there is a perfect match in both tables. A LEFT JOIN returns all the rows from the first (left) table, along with matching rows from the second table; if there is no match, it just displays NULL for the missing side.

10. The Importance of Constraints : Explain the purpose of the UNIQUE and NOT NULL constraint . How do they help ensure the quality 
        of data being entered into a database?.
           
           ANSWER : Constraints are rules enforced on data columns to block invalid data from entering the database,  
                    thereby ensuring data accuracy and reliability. The UNIQUE constraint guarantees that all values in a column are distinct from one another, preventing accidental duplicate accounts or emails. The NOT NULL constraint forces a field to always contain data, ensuring critical information like a user's password or ID can never be left completely blank
