```python
#12.Stats Function: Write a functions that accepts a list of numbers and returns the sum ans average of the list.

def cal_stats(num):
    #check if list is empty 
    if not num:
        return 0
    #calcualte the sum and average of a number
    total_sum = sum(num)
    average = total_sum / len(num)
    return total_sum, average

#list
my_list = [11,22,33,12,13,10,24]
total , avg = cal_stats(my_list)

print("sum:",total)
print("Average:",avg)
```

```python
#14.Manual palindrome: write a program to check whether a given string is a palindrome without using any built in reverse function.
#take a input from the user
t = input("Enter the string:")
print(t)
clean_text = t.lower()
#reconstruct the string manually
reversed_t = ""
for char in clean_text :
    reversed_t = char + reversed_t

#check if the original t matches the reversed text

if t == reversed_t:
    print(f"'{t}' is a palindrome.")
else:
    print(f"'{t}' is not a palindrome.")
```

2. NumPy Efficiency: How does Numpy internally handle large numerical data faster than python lists?

     ANSWERS : NumPy handles data much faster than standard Python lists due to how it organizes memory and handles operations under the hood:

           Block Memory (Contiguous Storage): A standard Python list doesn't actually hold values; it holds pointers scattered across memory that point to objects. NumPy arrays store actual data tightly together in a single, continuous block of memory. This allows your computer's processor to load and read the data almost instantly.
           
           Vectorization (No Hidden Loops): When you want to add numbers in Python, Python has to look up the data type of every single element one by one in a loop. NumPy uses pre-compiled C code to perform the operation on the entire data block at once at the hardware level, bypassing Python's slow internal overhead.
