PROJECT 1 : CONSOLE BASED SMART CALCULATOR 

```python
def calculator_operation():
    print("Welcome to  calculator!")
    print(" select an operation:")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Power")
    print("6. Modulus")

    operation = input("Enter the number  operation: ")

    if operation in ['1', '2', '3', '4', '5', '6'  ]:
        num1 = float(input("Enter the first number: "))
        num2 = float(input("Enter the second number: "))

        if operation == '1':
            result = num1 + num2
            print(f"The result of {num1} + {num2} is: {result}")
        elif operation == '2':
            result = num1 - num2
            print(f"The result of {num1} - {num2} is: {result}")
        elif operation == '3':
            result = num1 * num2
            print(f"The result of {num1} * {num2} is: {result}")
        elif operation == '4':
            if num2 != 0:
                result = num1 / num2
                print(f"The result of {num1} / {num2} is: {result}")
            else:
                print("Error: Division by zero is not allowed.")
        elif operation == '5':
            result = num1 ** num2
            print(f"The result of {num1} raised to the power of {num2} is: {result}")   
        elif operation == '6':
            if num2 != 0:
                result = num1 % num2
                print(f"The result of {num1} modulus {num2} is: {result}")
            else:
                print("Error: Modulus by zero is not allowed.")
    else:
        print("Invalid operation selected. Please try again.")
```

project 2 : STUDENTS MARKS ANALYZER

```python
import numpy as np

def students_grades():
    print(" Student Grades !")
    student1 = int(input("Enter the marks for student 1: "))
    student2 = int(input("Enter the marks for student 2: "))
    student3 = int(input("Enter the marks for student 3: "))

    grades = [student1, student2, student3]
    total_marks = np.sum(grades)
    print(f"The total marks for the class is: {total_marks}")
    highest_grade = np.max(grades)
    print(f"The highest grade in the class is: {highest_grade}")
    lowest_grade = np.min(grades)
    print(f"The lowest grade in the class is: {lowest_grade}")
    average_grade = np.mean(grades)
    print(f"The average grade for the class is: {average_grade:.2f}")
    if average_grade > 85:
        print("Grade: A")
    elif average_grade > 70:
        print("Grade: B")
    elif average_grade :
        print("Grade: C")
    elif average_grade >50:
        print("Grade: D")
    else:
        print("Grade: F")
```

PROJECT 3 DAILY EXPENSIVE TRACKER 

```python
import numpy as np
def daily_expenses():
    print("Daily Expenses Tracker!")
    expenses = int(input("Enter the number of expenses you want to track: "))
    expense_list = []
    for i in range(expenses):
        expense = float(input(f"Enter expense {i+1}: "))
        expense_list.append(expense)

    total_expenses = np.sum(expense_list)
    print(f"The total expenses for the day is: {total_expenses}")
    average_expense = np.mean(expense_list)
    print(f"The average expense for the day is: {average_expense:.2f}")
    highest_expense = np.max(expense_list)
    print(f"The highest expense for the day is: {highest_expense}")
    lowest_expense = np.min(expense_list)
    print(f"The lowest expense for the day is: {lowest_expense}")
```
